```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
df = pd.DataFrame(data=iris.data,columns=iris.feature_names)
df['species'] = iris.target
df['species'] = df['species'].map({0: 'setosa',1: 'versicolor',2: 'virginica'})
```

```python
df[['petal length (cm)', 'petal width (cm)']].cumsum().plot()
plt.title("Cumulative Petal Measurements Over Rows")
plt.show()
```

```python
df.groupby('species')['petal length (cm)'].mean().plot(kind='bar')
plt.title("Average Petal Length by Species")
plt.ylabel("Length (cm)")
plt.show()
```

```python
sns.histplot(df['petal length (cm)'], kde=True)
plt.title("Distribution of Petal Length")
plt.show()
```

```python
sns.scatterplot(x='petal length (cm)', y='petal width (cm)', hue='species', data=df)
plt.title("Petal Length vs Width by Species")
plt.show()
```
